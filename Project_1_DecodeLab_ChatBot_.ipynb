```python
import random
```

### Responses , for a bit of variety

```python
responses = {
    "hello" : "Hi there ! How can I help you today?" ,
    "hi "  :  "Hello ! Whats on your mind? " ,
    "hey" : "Hey! Good to see you." ,
    "how are you" : "I'm just a program , but I'm running smoothly! How about you?" ,
    "what is your name " : "I'm Steve , your friendly rule-based chatbot. " ,
    "who created you" : "I was built as part of the DecodeLabs AI Internship , Project 1." ,
    "what can you do" : "I can responed to a few predefined questions using simple rule based logic",
    "thank you" : "You're welcome!" ,
    "thanks" : "No problem at all!" ,
}
```

### Exit Commands

```python
exit_commands = {"bye" , "exit" , "quit"}
```

### Fallback Response(used when no rule matched )

```python
fallback_responses = [
    "I do not understand that yet. Could you rephrase?" ,
    "Sorry , I don't have a rule for that input." ,
    "Hmm , I'm not sure how to respond to that.",
]
```

### Sanitization function
                         Normalize user input so that "Hello" , "hello","HELLO", all map to the same dictionary key.

```python
def sanitize(raw_input):
  return raw_input.lower().strip()
```

## Response Function
                    (uses .get() ->atomic lookup + fallback)

```python
def get_response(clean_input):
  return responses.get(clean_input, random.choice(fallback_responses))
```

# Main Consversation Loop
                         (The Heartbeat / Infinite loop)

```python
def run_chatbot():
  print("Bot: Hi! I'm Steve. Type 'bye' anytime to end our chat.")

  while True :
    raw_input_text = input("You: ")
    clean_input = sanitize(raw_input_text)

    # Exit strategy : clean break out of the loop
    if clean_input in exit_commands:
      print("Bot: Goodbye! Have a great day.")
      break

    # Rule-based lookup with fallback
    reply = get_response(clean_input)
    print(f"Bot: {reply}")
# run the bot

if __name__ == "__main__":
  run_chatbot()
```
